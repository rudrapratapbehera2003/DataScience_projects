```python
"""
Problem Statement : Hospital Patient Data Analysis
Context:
A hospital maintains patient records including admission details, department, diagnosis, doctor, and bill amount.
You have two datasets: one with patient info and another with billing details. Some patients have blank bill amounts, 
and there are multiple rows for the same patient due to follow-ups.

Tasks:
1.	Load the patient dataset and show summary with info().
2.	Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount'].
3.	Drop administrative columns like ['ReceptionistID', 'CheckInTime'].
4.	Use groupby to find total bill amount per department.
5.	Remove duplicate patient records based on PatientID.
6.	Fill missing BillAmount values with the mean bill amount.
7.	Merge the billing dataset with patient dataset on PatientID.
8.	Concatenate an additional DataFrame that contains new patients for the current week (row-wise).
9.	Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).

Expected Outcome:
•	Final cleaned dataset with accurate billing info.
•	All missing values handled, merged dataset across PatientID.
•	Ability to perform further analytics on department-wise revenue or doctor performance.

"""
import pandas as pd 


# loading files to the enviroment
df1 = pd.read_csv("Patient_Data.csv")

# Loading billing dataset 
df2 = pd.read_csv("Billing_Data.csv")

# I am going to work on a copy to keep original intact
data_patient = df1.copy()


# 1.Load the patient dataset and show summary with info().
print(data_patient.info())


# 2.Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount'].
# 3.Drop administrative columns like ['ReceptionistID', 'CheckInTime'].
billing_patient = data_patient[['PatientID', 'Department', 'Doctor', 'BillAmount']]
print(billing_patient)

# 4.Use groupby to find total bill amount per department.
bill_amt_per_department = billing_patient.groupby("Department")["BillAmount"].sum().reset_index()
print(bill_amt_per_department)

# 5.Remove duplicate patient records based on PatientID.
print("\n<--------- Removing Duplicates From The Dataset According To The PatientID. -------> \n")
billing_patient.drop_duplicates(subset="PatientID",inplace=True)
print(billing_patient)

print("\n<------ Loking for how many empty cell are in different columns--------> \n")
print(billing_patient.isnull().sum())

# 6.Fill missing BillAmount values with the mean bill amount.
print("\n <------ finding the mean of the billmount to fill the missing values------>\n")
value = billing_patient["BillAmount"].mean()
print(f"The mean value is :{value}")

print("\n<--------Filling the missing values with the mean value of the bill amount -------->\n")
billing_patient["BillAmount"] = billing_patient["BillAmount"].fillna(value)
print(billing_patient)

# 7.Merge the billing dataset with patient dataset on PatientID.
print("\n<-------Merging------>\n")
merged_dataset = pd.merge(billing_patient,df2,on="PatientID")
print(merged_dataset)

```

```python
# 8.Concatenate an additional DataFrame that contains new patients for the current week (row-wise).
# 9. Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).
# Lets say an additional dataframe is given by 
patient_data_new_week = {
    "PatientID":[106,107,108,109,110],
    "Department":["Cardiology","Neurology","Orthopedics"," Cardiology","Dermatology "],
    "Doctor":["Mona","Suresh","Purnima","Vairava","Niranjan"],
    "BillAmount":[2000,3000,4000,5000, 6000]
}

patient_new_data = pd.DataFrame(patient_data_new_week)
print(patient_new_data)

# Lets create new billing that has been doene for these patients 
new_billings_week = {
    "PatientID":[106,107,108,109,110],
    "InsuranceCovered":[1000,1000,2000,2000,3000],
    "FinalAmount":[1000,2000,2000,3000,3000]
}
new_billing = pd.DataFrame(new_billings_week)
# Merging the two data frame first using the patientid

print("\n<---------Our new columns according to the neww weekday--------->\n")
merging_new = pd.merge(patient_new_data,new_billing, on="PatientID")
print(merging_new)

```

```python
# The Final output Where there is no vacant place and no duplicates is given buy 
df_with_new_patients = pd.concat([merged_dataset, merging_new], axis=0, ignore_index=True)
print(df_with_new_patients)
```
