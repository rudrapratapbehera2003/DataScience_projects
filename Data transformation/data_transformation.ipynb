````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Loading dataset into the enviroment
dataset = pd.read_csv("adult_with_headers (1).csv")

# Working on a copy data
df = dataset.copy()

# Understanding dataset
print("\n<------It will give the information about our data------->\n")
print(df.info())

print("\n<------It will describe all about our dataset------->\n")
print(df.describe(include="all"))

print("\n<------It will give the number of null values  in each columns------->\n")
print(df.isnull().sum())

print("\n<----------Correlation between the variables------------>\n")
print(df.corr(numeric_only=True))
````

````python
#Treating the inapropriate values
df = df.replace(" ?", np.nan)

# Filling missing values in categorical cols using median
for col in df.select_dtypes(include=['object']):
    df[col]= df[col].fillna(df[col].mode())

# Filling missing values in numerical cols using median
for col in df.select_dtypes(include=['int64']):
    df[col] = df[col].fillna(df[col].median())
    
````

````python
# Feature Engineering 

df["age_bucket"] = pd.cut(df["age"],bins=[0,30,45,60,100], labels=["Young","Mid_age","Senior","Old"])

# Capital Difference
df["capital_diff"] = df["capital_gain"] - df["capital_loss"]

# Log Transformation for skewed feature
df["capital_gain_loss"] = np.log1p(df['capital_gain'])
````

````python
# Encoding 
from sklearn.preprocessing import StandardScaler,OneHotEncoder, LabelEncoder,MinMaxScaler
#Identifying categorical cols
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# One-Hot Encoding (for <5 categories)
onehot_cols = [col for col in categorical_cols if df[col].nunique() <= 5]
# label Encoding (for >5 categorical)
label_cols = [col for col in categorical_cols if df[col].nunique() > 5]

#Applying one_hot
df = pd.get_dummies(df,columns=onehot_cols, drop_first=True) # Directly apply the on-Hot encoding

# Applying label Encoding 
le = LabelEncoder()
for col in label_cols:
    df[col] = le.fit_transform(df[col])
````

````python
# Scaling 

numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

#Using StandardScaler .It is bacically used for linear models
scaler_std = StandardScaler()
df_std= df.copy()
#Applying standard scaler
df_std[numerical_cols] = scaler_std.fit_transform(df_std[numerical_cols])


# Using Min-Max scaler, basically used for the tree structures like decision tree and so on.
scaler_mm = MinMaxScaler()
df_mm = df.copy()
# Applying min_max scaler
df_mm[numerical_cols] = scaler_mm.fit_transform(df_mm[numerical_cols])
````
