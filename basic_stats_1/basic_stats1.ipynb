````python
"""
Descriptive Analytics for Numerical Columns
●	Objective: To compute and analyze basic statistical measures for numerical columns in the dataset.
●	Steps:
	Load the dataset into a data analysis tool or programming environment (e.g., Python with pandas library).
	Identify numerical columns in the dataset.
	Calculate the mean, median, mode, and standard deviation for these columns.
	Provide a brief interpretation of these statistics.

"""
import pandas as pd

#Loading the data to the enviroment
df = pd.read_csv('sales_data_with_discounts.csv')

# Finding Numerical columns
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()

# finding mean for all the Numerical columns
print("<-------Finding Mean of all the numerical columns--------->\n")
for x in numerical_cols:
    print(f"Mean of the {x} column -->", df[x].mean())

print("\n<-------Finding the median of the all numerical columns--------->\n")
for x in numerical_cols:
    print(f"Median of the {x} column-->", df[x].median())


print("\n<-------Finding the mode of the all numerical columns--------->\n")
for x in numerical_cols:
    print(f"Mode of the {x} column -->", df[x].mode()[0])
````

## Brief Interpretation :

### Mean: 
#### mean represents the average of a numerical dataset. Mathematically, It is the sum of all the values divided by the number of values.
### Median: 
#### This is the middlle value of a numerical dataset.
#### Mode:
#### This is the most frequent value in a datset.



````python
"""
Data Visualization
●	Objective: To visualize the distribution and relationship of numerical and categorical variables in the dataset.
●	Histograms:
	Plot histograms for each numerical column.
	Analyze the distribution (e.g., skewness, presence of outliers) and provide inferences.
●	Box Plots:
	Create boxplots for numerical variables to identify outliers and the interquartile range.
	Discuss any findings, such as extreme values or unusual distributions.
●	Bar Chart Analysis for Categorical Column:
	Identify categorical columns in the dataset.
	Create bar charts to visualize the frequency or count of each category.
	Analyze the distribution of categories and provide insights.
Conclusion
●	Summarize the key findings from the descriptive analytics and data visualizations.

"""
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('sales_data_with_discounts.csv')

#Extracting Numerical columns
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()

# Extracting Categorical columns
categorical_cols = df.select_dtypes(exclude=["number","datetime64"]).columns.tolist()

for col in ["Date"]:
    if col in categorical_cols:
        categorical_cols.remove(col)
        
print(categorical_cols)


# For Numerical columns
for col in numerical_cols:
    plt.figure(figsize=(20,7))
    #Plotting histogram for each numerical columns
    plt.subplot(1,2,1)
    plt.title(f"Histogram of column --> {col}")
    sns.histplot(df[col], kde=True)


    #plotting the boxplot for the numerical columns
    plt.subplot(1,2,2)
    plt.title(f"Box plot of column ---> {col}")
    sns.boxplot(df[col])

    plt.show()



print("\n<--------Plotting Bar graphs for the categorical columns------->\n")
for col in categorical_cols:
    plt.figure(figsize=(20, 7))
    plt.title(f"Bar graph for column ----> {col}")
    counts = df[col].value_counts()
    plt.bar(counts.index, counts.values)

    plt.show()
````

### Key Findings By Analyzig Graphs
```
Histogram:

i. This plot shows, how the data has been distributed in our dataset. So, it's often known as distribution plot.

ii. In the above figure some plots tail have been draged right side,
which shows concentration of data points are less in the right side whereas the concentration of data points are high towards the lower values.
Hence, These plots are right-skewed distribution.
eg. This includes histplot of - Day,AvgPrice, Total sales Value, Discount Amount, Netsales

iii. In some cases some plots tail have been dragged towards left side,
which shows concentration of the data points are less in the left side whereas the concentration of data points are high towards higher values.
hence, This is known as left-skewed distribution.
eg. Discount rate(%)

Boxplot:

i. This plot is specially used to detect the outliers.
We can clearly see the points that are outside the lower whisker and the uppper whisker, these are nothing but outliers.

ii. Using boxplot we can also tell whether data has been distributed normally or their is skewness.

iii. If the box is in the middle portion and the middle line(indicates median) is in the center of that box theen we can clearly say that data has been distributed normally.

iv. If the box is closed towards the lower-whisker  and the middle line also towards the left side then we can say that data distribution is skewed towards right.  eg.  Discount Amount, Netsales

v. If the box is closed towards the upper-whisker and the middle line also towards the right side then we can say that data distribution is skewed towards left. eg. Discount(%)
```
