## 2. Calculate the Test Statistic:

````python


from  scipy.stats import norm

import math
#Given Data
x_mean = 3050 # The sample mean 
p_mean = 1000 + 5*600 # population mean 
std = 5*25 # Standar deviation
n = 25 # sample size

# Finding the z_score by using the formula
z_score = (x_mean - p_mean)/(std/math.sqrt(25))

print("\n<-------The z-score value is given by--------->\n")
print(z_score)

````

## 3. Determine the Probability and compare:

````python
# Finding the pobability 
p_value = 2*(1 - norm.cdf(abs(z_score)))
print(p_value)
````

## 4. Make a Decision:
```
Ans: As, We can see from the above calculation,
p_value < 0.05,
which is our significanc value.

Hence, We reject the Null Hypothesis.
so, The true mean weekly operating cost is not equal to theoratical model cost.
```
