Background
In quality control processes, especially when dealing with high-value items, destructive sampling is a necessary but costly method to ensure product quality. The test to determine whether an item meets the quality standards destroys the item, leading to the requirement of small sample sizes due to cost constraints.

Scenario
A manufacturer of print-heads for personal computers is interested in estimating the mean durability of their print-heads in terms of the number of characters printed before failure. To assess this, the manufacturer conducts a study on a small sample of print-heads due to the destructive nature of the testing process.

Data
A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29

## Answer
```

a. Build 99% Confidence Interval Using Sample Standard Deviation
Assuming the sample is representative of the population, construct a 99% confidence interval for 
the mean number of characters printed before the print-head fails using the sample standard deviation.
Explain the steps you take and the rationale behind using the t-distribution for this task.

```

````python

import math
import numpy as np

# List of durabilities of randomly selected different print heads.
durabilities_list = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29] # This is our sample

# Finding the sample mean 
s_mean = np.mean(durabilities_list)
print(s_mean)
````

## Challenges:
### Standard Deviation:
```
Here, The population standard deviation is unknown,
so we will go for the calculation of sample standard deviation.
```
### Finding t_score value:
##### Why we need t_score value ?
``` Ans: To calculate the confidence interval. ```
## Why chossing t-distribution instead of the z?
```
Ans: Because,
    i) The sample data is less then 30.
    ii) The population standard deviation is not given.
```

````python
# Finding standard deviation 
std = np.std(durabilities_list,ddof=1)
print(std)
````

````python
# Finding t_score and p_value

from scipy.stats import t
#Data we have 

n = len(durabilities_list)
df = n-1 # Degrees of freedom
alpha = 0.01 # significance value as it is given 99% so using formula 1-0.99 = 0.01

# t_value
t_value_div2 = t.ppf(1-(alpha/2), df)

print("Degrees of freedom :",df)
print("t_value is :",t_value_div2)
````

## Finding Confidence Interval

````python
# C.I_lower value
ci_lower = s_mean - (t_value_div2 * (std / math.sqrt(n)))
print(ci_lower)

ci_upper = s_mean + (t_value_div2 * (std / math.sqrt(n)))
print(ci_upper)
````

```
b. Build 99% Confidence Interval Using Known Population Standard Deviation
If it were known that the population standard deviation is 0.2 million characters, construct a 99% confidence interval for the mean number of characters printed before failure.
```

````python
# IF we know the population standard deviation we will go for the z_distribution
from scipy.stats import norm
#Data we have here 
p_std = 0.2 # population standard deviation
z_value_div2 = norm.ppf(1-(alpha/2))

# Calculating the ci
ci_lower_new = s_mean - z_value_div2 * (p_std / math.sqrt(n))
ci_upper_new = s_mean + z_value_div2 * (p_std / math.sqrt(n))

print(ci_lower_new)
print(ci_lower_new)
````
